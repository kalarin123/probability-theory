$n = A_6^3 = 6 \cdot 5 \cdot 4$

$P(A) = \frac{1}{A_6^3} = \frac{1}{120}$

Все 6 карточек можно переставить $6!$ способами. Благоприятный исход один:

$P(B) = \frac{1}{6!} = \frac{1}{720}$

```python
round(1 / sc.factorial(6), 4)
```

Общее число способов выбрать 7 мест из 36:

Два купе из девяти выбираются $C_9^2$ способами. Из 8 мест выбранных купе нужно выбрать 7:

$m = C_9^2 \cdot C_8^7$

$P(A) = \frac{C_9^2 \cdot C_8^7}{C_{36}^7}$

```python
round(sc.comb(9, 2) * sc.comb(8, 7) / sc.comb(36, 7), 7)
```

* Рассмотрим событие $B$ – пассажиры попали в **семь купе**.

Семь купе из девяти выбираются $C_9^7$ способами. В каждом купе выбирается одно из четырех мест:

$m = C_9^7 \cdot 4^7$

$P(B) = \frac{C_9^7 \cdot 4^7}{C_{36}^7}$

```python
round(sc.comb(9, 7) * 4 ** 7 / sc.comb(36, 7), 5)
```

* Рассмотрим событие $D$ – пассажиры попали в **три купе**.

Три купе из девяти выбираются $C_9^3$ способами.

$4+2+1,\quad 3+3+1,\quad 3+2+2$

Поэтому число благоприятных случаев:

$m = C_9^3 \left(6C_4^4C_4^2C_4^1 + 3C_4^3C_4^3C_4^1 + 3C_4^3C_4^2C_4^2\right)$

$P(D) = \frac{m}{C_{36}^7}$

```python
m = sc.comb(9, 3) * (
    6 * sc.comb(4, 4) * sc.comb(4, 2) * sc.comb(4, 1)
    + 3 * sc.comb(4, 3) * sc.comb(4, 3) * sc.comb(4, 1)
    + 3 * sc.comb(4, 3) * sc.comb(4, 2) * sc.comb(4, 2)
)
round(m / sc.comb(36, 7), 5)
```

### Задача 1.38
> Пятитомное собрание сочинений расположено на полке в случайном порядке. Какова вероятность того, что книги стоят слева направо в порядке нумерации томов (от 1 до 5)?

Рассмотрим событие $A$ — книги стоят в порядке $1,2,3,4,5$.

Всего существует

$P_5 = 5!$

перестановок пяти томов. Благоприятный вариант только один, поэтому

$P(A)=\frac{1}{5!}=\frac{1}{120}\approx0.00833$.

```python
orders = list(permutations(range(1, 6)))
favorable = [order for order in orders if order == (1, 2, 3, 4, 5)]

p = len(favorable) / len(orders)

print(f'Всего перестановок: {len(orders)}')
print(f'Благоприятных: {len(favorable)}')
print(f'P(A) = {p:.5f}')
```

### Задача 1.43
> Наудачу взятый телефонный номер состоит из 5 цифр. Какова вероятность того, что в нем все цифры: а) различные; б) одинаковые; в) нечетные? Известно, что номер телефона не начинается с цифры ноль.

Всего допустимых пятизначных телефонных номеров:

$n=9\cdot10^4=90000$.

**а)** Все цифры различные. Для первой цифры — 9 вариантов, далее соответственно 9, 8, 7 и 6 вариантов:

$m_a=9\cdot9\cdot8\cdot7\cdot6=27216$,

$P(A)=\frac{27216}{90000}=0.3024$.

**б)** Все цифры одинаковые. Подходят номера $11111,22222,\ldots,99999$:

$m_b=9$,

$P(B)=\frac{9}{90000}=0.0001$.

**в)** Все цифры нечетные. Для каждой позиции возможны цифры $1,3,5,7,9$:

$m_c=5^5=3125$,

$P(C)=\frac{3125}{90000}=\frac{5}{144}\approx0.0347$.

```python
digits = '0123456789'
first_digits = '123456789'

numbers = [
    (first,) + rest
    for first in first_digits
    for rest in product(digits, repeat=4)
]

total = len(numbers)

different = sum(len(set(number)) == 5 for number in numbers)
same = sum(len(set(number)) == 1 for number in numbers)
odd = sum(all(int(digit) % 2 == 1 for digit in number) for number in numbers)

print(f'Всего допустимых номеров: {total}')
print(f'а) Все цифры различные: {different} случаев, P = {different / total:.4f}')
print(f'б) Все цифры одинаковые: {same} случаев, P = {same / total:.4f}')
print(f'в) Все цифры нечетные: {odd} случаев, P = {odd / total:.4f}')
```

# Лабораторная работа № 3. Вероятности сложных событий. Принцип включений–исключений. Беспорядки

**Цель:** применять формулы дополнения, суммы и произведения вероятностей, подсчитывать объединения множеств и перестановки без неподвижных точек.

**Основные формулы:**

- $P(A\cup B)=P(A)+P(B)-P(A\cap B)$.
- $P(A\cup B\cup C)=P(A)+P(B)+P(C)-P(A\cap B)-P(A\cap C)-P(B\cap C)+P(A\cap B\cap C)$.
- $P(\bar A)=1-P(A)$.
- При независимости: $P(A\cap B)=P(A)P(B)$.
- Число беспорядков: $D_n=n!\sum_{k=0}^{n}(-1)^k/k!$.

У каждой задачи ниже отдельно приведено математическое решение и проверка числа с помощью Python.

```python
from math import comb, factorial
from fractions import Fraction
from itertools import permutations

def D(n):
    """Число беспорядков (перестановок без неподвижных точек)."""
    return sum((-1)**k * comb(n, k) * factorial(n-k) for k in range(n+1))

assert [D(k) for k in range(7)] == [1, 0, 1, 2, 9, 44, 265]
```

### Задача 1.58. Три контрольные работы

**Условие.** Вероятности своевременного выполнения трёх работ: $p_1=0{,}6$, $p_2=0{,}5$, $p_3=0{,}8$. Найти вероятность выполнения **ровно двух** и **не менее двух** работ. При вычислении считаем события независимыми.

**Решение.** Обозначим $q_1=0{,}4$, $q_2=0{,}5$, $q_3=0{,}2$. Ровно две работы выполнены в одном из трёх несовместных случаев:

$$P_2=p_1p_2q_3+p_1q_2p_3+q_1p_2p_3$$
$P_2=0{,}6\cdot0{,}5\cdot0{,}2+0{,}6\cdot0{,}5\cdot0{,}8+0{,}4\cdot0{,}5\cdot0{,}8=0{,}46.$$

Все три работы выполнены с вероятностью $P_3=0{,}6\cdot0{,}5\cdot0{,}8=0{,}24$.

Поэтому $P(\text{не менее двух})=P_2+P_3=0{,}46+0{,}24=0{,}70$.

**Ответ:** а) $0{,}46$; б) $0{,}70$.

```python
p1,p2,p3 = Fraction(6,10),Fraction(5,10),Fraction(8,10)
q1,q2,q3 = 1-p1,1-p2,1-p3
exactly_two = p1*p2*q3 + p1*q2*p3 + q1*p2*p3
at_least_two = exactly_two+p1*p2*p3
print('1.58: ровно 2 =', exactly_two, '; не менее 2 =', at_least_two)
assert (exactly_two,at_least_two)==(Fraction(23,50),Fraction(7,10))
```

### Задача 1.67. Минимальное число испытаний

**Условие.** Вероятность наступления события в одном испытании $p=0{,}4$. Сколько независимых испытаний требуется, чтобы вероятность хотя бы одного успеха была не меньше $0{,}8704$?

**Решение.** Вероятность не получить событие в одном испытании $q=0{,}6$, в $n$ независимых испытаниях — $q^n$. Требуем:

$$1-0{,}6^n\geq0{,}8704 \quad\Longleftrightarrow\quad 0{,}6^n\leq0{,}1296.$$

Поскольку $0{,}6^4=0{,}1296$, минимальное $n=4$. При $n=3$ вероятность меньше требуемой.

**Ответ:** $4$ испытания.

```python
p=Fraction(4,10); target=Fraction(8704,10000)
n_167=next(n for n in range(1,100) if 1-(1-p)**n>=target)
print('1.67: n =',n_167);assert n_167==4
```

### Задача 3.1. Сданные экзамены

**Условие.** Из $100$ студентов историю сдали $56$, математику $44$, физику $42$; историю или математику — $76$, историю или физику — $76$, математику или физику — $66$; все три — $5$. Найти: а) сдавших хотя бы один предмет; б) не сдавших ни одного; в) сдавших только физику.

**Решение.** Из двухмножественной формулы получаем $|И\cap М|=56+44-76=24$, $|И\cap Ф|=56+42-76=22$, $|М\cap Ф|=44+42-66=20$.

$$|И\cup М\cup Ф|=56+44+42-24-22-20+5=81.$$

Не сдали ничего $100-81=19$. Только физику $42-22-20+5=5$.

**Ответ:** а) $81$; б) $19$; в) $5$.

```python
hm,hf,mf=56+44-76,56+42-76,44+42-66
at_least_one=56+44+42-hm-hf-mf+5
none=100-at_least_one
only_physics=42-hf-mf+5
print('3.1:',at_least_one,none,only_physics)
assert (at_least_one,none,only_physics)==(81,19,5)
```

# Лабораторная работа № 4
## Условная вероятность и формула полной вероятности

**Тема:** классическое решение и моделирование задачи с двумя урнами.

**Цель:** вычислить вероятность по формуле полной вероятности и проверить её методом Монте-Карло.

**Условие.** В первой урне 15 белых и 5 чёрных шаров, во второй — 6 белых и 1 чёрный. Из первой урны случайно вынимают 3 шара **без возвращения** и перекладывают во вторую. Затем из второй урны случайно вынимают 1 шар. Найти вероятность того, что последний шар белый.

### 1. Классическое решение

Пусть $A$ — событие «из второй урны вынут белый шар»; $H_k$ — гипотеза «из первой во вторую урну перенесено ровно $k$ белых шаров», $k=0,1,2,3$.

Всего способов выбрать 3 шара из первой урны: $C_{20}^{3}=1140$.

Вероятность гипотезы $H_k$ (гипергеометрическое распределение):

$$P(H_k)=\frac{C_{15}^{k}C_{5}^{3-k}}{C_{20}^{3}}.$$

После переноса во второй урне будет 10 шаров, среди которых $6+k$ белых. Поэтому

$$P(A\mid H_k)=\frac{6+k}{10}.$$

По **формуле полной вероятности**:

$$P(A)=\sum_{k=0}^{3}P(H_k)P(A\mid H_k).$$

```python
from math import comb
from fractions import Fraction

ways_total = comb(20, 3)
probability = Fraction(0, 1)

print('Перенесено белых | Число способов | P(H_k) | P(A | H_k)')
for k in range(4):
    ways = comb(15, k) * comb(5, 3-k)
    p_h = Fraction(ways, ways_total)
    p_white_given_h = Fraction(6+k, 10)
    probability += p_h * p_white_given_h
    print(f'{k:^17} | {ways:^13} | {str(p_h):^8} | {p_white_given_h}')

print(f'\nТочная вероятность: {probability} = {float(probability):.6f}')
assert probability == Fraction(33, 40)
```

Подставим все четыре гипотезы:

$$P(A)=\frac{10}{1140}\cdot\frac6{10}+\frac{150}{1140}\cdot\frac7{10}+\frac{525}{1140}\cdot\frac8{10}+\frac{455}{1140}\cdot\frac9{10}$$

$$=\frac{60+1050+4200+4095}{11400}=\frac{9405}{11400}=\frac{33}{40}=0{,}825.$$

**Ответ по классической формуле: $P(A)=0{,}825=82{,}5\%$.**

### 2. Проверка классического решения другим способом

После переноса во второй урне 10 шаров. Из них 7 первоначально находились во второй урне и 3 были перенесены из первой. Если вытянут шар из первоначальной части, шанс белого $6/7$; если перенесённый — $15/20$.

$$P(A)=\frac7{10}\cdot\frac67+\frac3{10}\cdot\frac{15}{20}=\frac6{10}+\frac9{40}=\frac{33}{40}.$$

```python
check = Fraction(7, 10)*Fraction(6, 7) + Fraction(3, 10)*Fraction(15, 20)
print('Аналитическая проверка:', check, '=', float(check))
assert check == probability
```

### 3. Моделирование методом Монте-Карло

Многократно воспроизводим условие задачи. В каждом испытании заново создаём обе урны, случайно перекладываем **три разных** шара из первой во вторую и выбираем один шар из второй. Отношение числа белых шаров к количеству испытаний даёт экспериментальную оценку вероятности.

Используем фиксированное начальное состояние генератора, чтобы результат воспроизводился. Сравним оценку после 100, 1000, 10 000 и 100 000 испытаний.

```python
import random

rng = random.Random(20260928)
checkpoints = (100, 1_000, 10_000, 100_000)
white_outcomes = 0
estimates = []

for experiment in range(1, checkpoints[-1] + 1):
    first_urn = ['белый'] * 15 + ['чёрный'] * 5
    second_urn = ['белый'] * 6 + ['чёрный']

    transferred = rng.sample(first_urn, 3)  # без возвращения
    second_urn.extend(transferred)
    drawn = rng.choice(second_urn)
    if drawn == 'белый':
        white_outcomes += 1

    if experiment in checkpoints:
        estimate = white_outcomes / experiment
        estimates.append(estimate)
        print(f'{experiment:>6} испытаний: P ≈ {estimate:.5f}; '
              f'отклонение = {abs(estimate - float(probability)):.5f}')

assert abs(estimates[-1] - float(probability)) < 0.01
```

### 4. Сравнение результатов и вывод

**Теоретическая вероятность:** $33/40=0{,}825$.

**Модельная вероятность:** при 100 000 испытаний получено $0{,}82404$; при другом начальном состоянии генератора оценка немного изменится.

**Вывод:** классический расчёт и компьютерное моделирование согласуются. При большом числе независимых повторений относительная частота белого шара близка к вероятности $0{,}825$.

# Лабораторная работа № 5. Повторные испытания в схеме Бернулли

## § 2.4. Повторные независимые испытания

В схеме Бернулли проводят $n$ независимых испытаний. Вероятность успеха в каждом испытании равна $p$, а неуспеха — $q=1-p$.

**Формула Бернулли:**

$$P_n(m)=C_n^m p^m q^{n-m},\quad C_n^m=\frac{n!}{m!(n-m)!}.$$

**Формула Пуассона** (для большого $n$ и малого $p$, где $\lambda=np$):

$$P_n(m)\approx\frac{\lambda^m e^{-\lambda}}{m!}.$$

**Локальная формула Муавра–Лапласа** (при достаточно большом $npq$):

$$P_n(m)\approx\frac{\varphi(x)}{\sqrt{npq}},\quad x=\frac{m-np}{\sqrt{npq}},\quad\varphi(x)=\frac{e^{-x^2/2}}{\sqrt{2\pi}}.$$

**Интегральная формула Муавра–Лапласа** позволяет найти вероятность попадания числа успехов в интервал. В Python используем `norm.cdf` — функцию распределения стандартной нормальной величины. Для включённых целочисленных границ можно применить поправку $0{,}5$.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import binom, poisson, norm
from math import sqrt
```

### Задача 2.9
> В среднем 20% пакетов акций на аукционах продаются по первоначально заявленной цене. Найти вероятность того, что из 9 пакетов акций по первоначальной цене: 1) не будут проданы 5 пакетов; 2) будет продано: а) менее 2; б) не более 2; в) хотя бы 2; г) наивероятнейшее число пакетов.

**Дано:** $n=9$, $p=0{,}2$, $q=0{,}8$.

**1)** Если 5 пакетов не проданы, значит проданы 4. По формуле Бернулли:

$$P_9(4)=C_9^4\cdot0{,}2^4\cdot0{,}8^5\approx0{,}06606.$$

**2а)** Менее двух проданных — это 0 или 1:

$$P(X<2)=P_9(0)+P_9(1).$$

**2б)** Не более двух — это 0, 1 или 2:

$$P(X\le2)=P_9(0)+P_9(1)+P_9(2).$$

**2в)** Хотя бы два — используем противоположное событие:

$$P(X\ge2)=1-P(X<2).$$

**2г)** Наивероятнейшее число определяем по неравенству

$$np-q\le m_0\le np+p,\qquad1\le m_0\le2.$$

Значит, наиболее вероятны **1 и 2 проданных пакета**. Вероятности этих исходов одинаковы.

```python
# Задача 2.9 — формула Бернулли
n_29, p_29 = 9, 0.2

p_29_1 = binom.pmf(4, n_29, p_29)
p_29_a = binom.cdf(1, n_29, p_29)
p_29_b = binom.cdf(2, n_29, p_29)
p_29_v = binom.sf(1, n_29, p_29)
p_29_m1 = binom.pmf(1, n_29, p_29)
p_29_m2 = binom.pmf(2, n_29, p_29)

print(f'1) Не проданы 5 пакетов: {p_29_1:.6f}')
print(f'2а) Менее 2 продано:    {p_29_a:.6f}')
print(f'2б) Не более 2:        {p_29_b:.6f}')
print(f'2в) Хотя бы 2:         {p_29_v:.6f}')
print('2г) Наивероятнейшее число: 1 и 2')
print(f'    P(X=1)={p_29_m1:.6f}; P(X=2)={p_29_m2:.6f}')
print(f'    Сумма вероятностей: {(p_29_m1+p_29_m2):.6f}')

assert np.isclose(p_29_a + p_29_v, 1)
assert np.isclose(p_29_m1, p_29_m2)
```

**Ответ:** 1) $0{,}06606$; 2) а) $0{,}43621$; б) $0{,}73820$; в) $0{,}56379$; г) $m_0=1$ и $m_0=2$ (по $0{,}30199$), суммарно $0{,}60398$.

$$P(480\le X\le520)\approx\Phi\left(\frac{520-500}{\sqrt{250}}\right)-\Phi\left(\frac{480-500}{\sqrt{250}}\right).$$

В обоих интервальных расчётах дополнительно проверяем результат с поправкой на непрерывность: для включённых границ $[a,b]$ берём $[a-0{,}5,b+0{,}5]$.

```python
# Задача 2.11 — локальная и интегральная теоремы Муавра–Лапласа
n_211, p_211 = 1000, 0.5
mu_211 = n_211 * p_211
sigma_211 = sqrt(n_211 * p_211 * (1-p_211))

# а) Ровно 480
approx_a_211 = norm.pdf((480-mu_211)/sigma_211) / sigma_211
exact_a_211 = binom.pmf(480, n_211, p_211)

# б) Наивероятнейшее число 500
approx_b_211 = norm.pdf(0) / sigma_211
exact_b_211 = binom.pmf(500, n_211, p_211)

# в) Не менее 480
approx_v_211 = norm.sf((480-mu_211)/sigma_211)
corrected_v_211 = norm.sf((479.5-mu_211)/sigma_211)
exact_v_211 = binom.sf(479, n_211, p_211)

# г) От 480 до 520 включительно
approx_g_211 = norm.cdf((520-mu_211)/sigma_211) - norm.cdf((480-mu_211)/sigma_211)
corrected_g_211 = norm.cdf((520.5-mu_211)/sigma_211) - norm.cdf((479.5-mu_211)/sigma_211)
exact_g_211 = binom.cdf(520, n_211, p_211) - binom.cdf(479, n_211, p_211)

print(f'а) 480:       приближённо {approx_a_211:.6f}; точно {exact_a_211:.6f}')
print(f'б) m₀ = 500: приближённо {approx_b_211:.6f}; точно {exact_b_211:.6f}')
print(f'в) ≥ 480:    без поправки {approx_v_211:.6f}; с поправкой {corrected_v_211:.6f}; точно {exact_v_211:.6f}')
print(f'г) 480–520:  без поправки {approx_g_211:.6f}; с поправкой {corrected_g_211:.6f}; точно {exact_g_211:.6f}')

assert abs(exact_a_211 - approx_a_211) < 0.0001
assert abs(exact_v_211 - corrected_v_211) < abs(exact_v_211 - approx_v_211)
assert abs(exact_g_211 - corrected_g_211) < abs(exact_g_211 - approx_g_211)
```

**Ответ по приближённым формулам из учебника:** а) $0{,}0113$; б) $500$ предприятий, вероятность $0{,}0252$; в) $0{,}897$; г) $0{,}794$. В коде также показаны точные значения и нормальные приближения с поправкой на непрерывность.
